## Q4. Linspace, Properties and Transpose

```
import numpy as np

Prabhjot = np.linspace(10, 100, 25)

print("Array:", Prabhjot)
print("Dimensions:", Prabhjot.ndim)
print("Shape:", Prabhjot.shape)
print("Total elements:", Prabhjot.size)
print("Data type:", Prabhjot.dtype)
print("Total bytes:", Prabhjot.nbytes)

# Transpose using reshape()
transpose = Prabhjot.reshape(25, 1)
print("Transpose using reshape():")
print(transpose)

# Yes, T can also be used.
print("Transpose using T:")
print(Prabhjot.reshape(1, 25).T)
```
