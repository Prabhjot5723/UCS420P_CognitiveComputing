```python
my_dict = {
    "name": "Prabhjot Singh",
    "roll_no": "1024170008",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}

print("Original Dictionary:", my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")
print("After Renaming:", my_dict)
```

```python
my_dict["cgpa"] = 8.5
print("After Adding CGPA:", my_dict)
```

```python
my_dict["age"] += 1
print("After Updating Age:", my_dict)
```

```python
d1 = my_dict.copy()
print("Removed:", d1.pop("branch"))
print("Using pop():", d1)
```

```python
d2 = my_dict.copy()
del d2["branch"]
print("Using del:", d2)
```

```python
for key, value in my_dict.items():
    print(key, "->", value)
```

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key not found.")

```

```python

friend_dict = {
    "name": "Rahul",
    "roll_no": "1024179999",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

merged = {**my_dict, **friend_dict}
print("Merged Dictionary:", merged)
```

```python
new_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("String Values Dictionary:", new_dict)
```
