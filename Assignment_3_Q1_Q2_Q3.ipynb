## Q1. Create the dataset

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma", "Frank", "Grace", "Harry", "Ivy"],
    "Age": [25, 30, 35, 28, 22, 40, 27, 33, 29],
    "City": ["Delhi", "Mumbai", "Chandigarh", "Pune", "Jaipur", "Kolkata", "Ludhiana", "Amritsar", "Patiala"],
    "Salary": [40000, 50000, 60000, 45000, 35000, 70000, 42000, 55000, 48000],
    "Experience": [2, 5, 8, 3, 1, 12, 4, 7, 6]
}

df = pd.DataFrame(data)
print("Q1. Dataset:")
display(df)
```

## Q2. Locate rows 0, 4, 7 and 8

```python
print("Q2. Rows 0, 4, 7 and 8:")
display(df.loc[[0, 4, 7, 8]])
```

## Q3. Navigate the DataFrame

```python
# Q3.1 Select rows from index 3 to 7
print("Q3.1 Rows from index 3 to 7:")
display(df.loc[3:7])

# Q3.2 Select rows from index 4 to 8 and column 2 to 4
print("Q3.2 Rows from index 4 to 8 and columns 2 to 4:")
display(df.iloc[4:9, 2:5])

# Q3.3 Select all rows with column index 1 to 3
print("Q3.3 All rows with column index 1 to 3:")
display(df.iloc[:, 1:4])
```
