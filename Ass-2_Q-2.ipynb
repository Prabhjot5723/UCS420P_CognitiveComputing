```python
L = [10, 0, 20, 40, 10, 70, 0, 0, 0, 80]
scores = tuple(L[:8])

print("Tuple:", scores)
```

```python
print("Highest:", max(scores))
print("Highest Index:", scores.index(max(scores)))
print("Lowest:", min(scores))
print("Lowest Count:", scores.count(min(scores)))
```

```python
rev = list(scores[::-1])
print("Reversed List:", rev)
```

```python
x = int(input("Enter score: "))
if x in scores:
    print("Index:", scores.index(x))
else:
    print("Not Present")
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
```
