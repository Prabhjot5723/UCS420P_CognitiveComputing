## Q5. Statistics, Reshape and Resize

```
import numpy as np

ucs420_Prabhjot = np.array([
    [10,20,30,40],
    [50,60,70,80],
    [90,15,20,35]
])

print("Mean:", np.mean(ucs420_Prabhjot))
print("Median:", np.median(ucs420_Prabhjot))
print("Max:", np.max(ucs420_Prabhjot))
print("Min:", np.min(ucs420_Prabhjot))
print("Unique elements:", np.unique(ucs420_Prabhjot))

reshaped_ucs420_Prabhjot = ucs420_Prabhjot.reshape(4,3)
print("Reshaped array:")
print(reshaped_ucs420_Prabhjot)

resized_ucs420_Prabhjot = np.resize(ucs420_Prabhjot, (2,3))
print("Resized array:")
print(resized_ucs420_Prabhjot)
```
