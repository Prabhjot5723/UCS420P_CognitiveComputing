```python
roll = 1024170008
```

```python
digits = [int(i) for i in str(roll)]
print("Digits:", digits)
```

```python
L = [i * 10 for i in digits]
print("List L:", L)
```

```python
print(L)
```

```python
L.append(90)
print("After append:", L)
```

```python
L.insert(2, 50)
print("After insert:", L)
```

```python
L.remove(0)
print("After remove:", L)
```

```python
L.pop(3)
print("After pop:", L)
```

```python
L.sort()
print("Ascending:", L)
```

```python
L.sort(reverse=True)
print("Descending:", L)
```

```python
print("First 3:", L[:3])
print("Last 3:", L[-3:])
```

```python
avg = sum(L) / len(L)
new_list = [x for x in L if x > avg]
print("Average:", avg)
print("New List:", new_list)
```
